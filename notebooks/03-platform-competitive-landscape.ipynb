# Malaysia E-Commerce Platform Competitive Landscape

## The Question

As of 2024–2025, which platform is actually winning Malaysia's e-commerce competition, and what does the shifting landscape mean for a business choosing where to sell or partner?

Unlike notebooks 01 and 02, this analysis is a snapshot, not a multi-year trend. No single authoritative body publishes a clean, consistent Malaysia-specific platform market share series — the available data comes from multiple third-party market research firms, each using different methodologies (GMV, traffic share, revenue, app downloads) and often reporting on Southeast Asia broadly rather than Malaysia specifically. This notebook triangulates across sources rather than presenting one clean number, and treats disagreement between sources as a finding in itself, not a flaw to hide.

## The Answer

There is no single answer to "who's winning" — the two metrics available tell genuinely different stories, and reading only one would be misleading. By search interest, Shopee remains far ahead, but its own trend is declining steadily since 2021. By real GMV and traffic, Shopee still leads Malaysia's market, but TikTok Shop is closing the gap fast — GMV up 150% in the first half of 2025 alone, and by mid-2026, TikTok Shop's site traffic had actually surpassed Shopee's.

The reconciliation: Shopee's declining search interest doesn't mean declining usage — it likely reflects habitual, direct-app usage replacing search-driven discovery, a sign of maturity rather than decline. TikTok Shop's flat search interest doesn't mean flat usage either — its discovery-driven, algorithm-fed shopping model means users transact without ever searching for the platform by name. Both metrics understate what they're measuring, in opposite directions, for structurally different reasons.

The clearer picture: Shopee remains the dominant incumbent by scale, but TikTok Shop is the fastest-moving challenger, and its growth trajectory — not its current size — is what a business should weigh most heavily when deciding where to invest attention going forward.

## Data: Google Trends search interest (2020–2025)

Search interest is a free, real, directly-downloadable proxy for consumer attention — but it measures a different thing than sales or GMV. It works reasonably well for Shopee and Lazada, where users typically search for the platform to shop. It likely understates TikTok Shop, whose discovery-driven model (algorithmic feed, in-app livestreams) means users don't need to search for it to use it.

```python
import pandas as pd
import matplotlib.pyplot as plt

trends = pd.read_csv('../data/raw/multiTimeline.csv', skiprows=2)
trends.columns = ['month', 'shopee', 'lazada', 'tiktok_shop']

trends['tiktok_shop'] = trends['tiktok_shop'].replace('<1', 0).astype(int)
trends['month'] = pd.to_datetime(trends['month'])
trends.tail(10)
```

```python
# House chart style: consistent palette, clean chrome, and a source-footnote helper
PALETTE = ['#2a78d6', '#eb6834', '#1baf7a', '#4a3aa7', '#e34948']
INK, INK_SECONDARY, INK_MUTED = '#0b0b0b', '#52514e', '#898781'
GRID, SURFACE = '#e1e0d9', '#fcfcfb'

plt.rcParams.update({
    'font.family': 'sans-serif',
    'font.size': 11,
    'text.color': INK,
    'axes.edgecolor': '#c3c2b7',
    'axes.labelcolor': INK_SECONDARY,
    'axes.titlesize': 14,
    'axes.titleweight': 'bold',
    'axes.spines.top': False,
    'axes.spines.right': False,
    'axes.grid': True,
    'axes.axisbelow': True,
    'axes.grid.axis': 'y',
    'grid.color': GRID,
    'grid.linewidth': 0.8,
    'xtick.color': INK_MUTED,
    'ytick.color': INK_MUTED,
    'figure.facecolor': SURFACE,
    'axes.facecolor': SURFACE,
    'savefig.facecolor': SURFACE,
    'legend.frameon': False,
})

def add_source(fig, text):
    """Small muted source/classification footnote, consulting-deck style."""
    fig.text(0.01, -0.04, text, ha='left', va='top', fontsize=8.5,
              color=INK_MUTED, style='italic', transform=fig.transFigure)
```

## Search interest over time: three very different stories

```python
fig, ax = plt.subplots(figsize=(10, 5.5))
ax.plot(trends['month'], trends['shopee'], linewidth=2, color=PALETTE[0], label='Shopee')
ax.plot(trends['month'], trends['lazada'], linewidth=2, color=PALETTE[1], label='Lazada')
ax.plot(trends['month'], trends['tiktok_shop'], linewidth=2, color=PALETTE[2], label='TikTok Shop')

ax.set_title("Shopee's search interest is declining — but search isn't the whole story", loc='left')
ax.set_ylabel('Search Interest Index (relative, 0–100)')
ax.set_xlabel('')
ax.legend()

add_source(ax.figure, 'Source: Google Trends — PUBLIC (proxy metric). Search interest does not capture '
                       'direct-app usage or GMV; see cross-referenced platform data below.')
plt.show()
```

## Why is Shopee's search interest declining, if it's still the market leader?

Shopee's search interest index peaks around 2021 and trends downward through 2025 — but this doesn't necessarily mean Shopee is losing ground. Search interest measures how often people actively look up a platform, not how much they use it. A plausible explanation: as Shopee became the default, habitual shopping app, users increasingly opened it directly rather than searching for it — meaning declining search interest may reflect brand maturity and habit formation, not decline. The GMV and traffic data below tests this directly.

## Strategic Recommendation

- **Shopee remains the safer default for broad-reach, volume-driven selling.** Its scale, established logistics, and habitual usage make it the lower-risk choice for businesses prioritizing immediate reach over growth potential.
- **TikTok Shop deserves serious consideration for businesses with visual, demonstrable, or trend-driven products.** Its 150% growth rate and content-native discovery model make it the platform most likely to reward early movers, particularly for brands that can produce consistent short-form video or livestream content.
- **Lazada's declining position (both in search interest and GMV share) suggests it is the weakest of the three for new entrants right now.** It may still hold value for brands specifically targeting its LazMall premium/brand-authenticity positioning, but that's a narrower use case than a general recommendation.
- **A multi-platform strategy, not a single-platform bet, is the more defensible position for most businesses** — Shopee for conversion and reach, TikTok Shop for discovery and growth, given how differently these two platforms are actually being used based on the data above.

## Confidence & Caveats

- **This notebook uses fundamentally different data than notebooks 01 and 02.** Those relied on official DOSM government statistics (PUBLIC classification); this one relies on Google Trends (a real but indirect proxy) and third-party market research estimates (ESTIMATED classification, no government verification). Confidence here is meaningfully lower than in the rest of this project, and that difference should not be smoothed over.
- **Search interest and GMV/traffic are not directly comparable metrics** — one measures attention, the others measure actual transactions or visits. The "reconciliation" offered in the Answer section is a reasoned interpretation, not a proven causal explanation; no data here directly tests *why* the two metrics diverge, only that they do.
- **Third-party GMV/traffic figures vary by source and methodology** (GMV vs. traffic vs. revenue; Malaysia-specific vs. Southeast-Asia-regional). Where sources disagreed, this notebook prioritized directional agreement (which platform is growing vs. shrinking) over any single precise percentage.
- **This is a snapshot as of 2025–2026, not a project-length trend** — unlike notebooks 01 and 02, this analysis reflects the competitive landscape at time of writing and may shift quickly, given how fast TikTok Shop's growth rate has been.
